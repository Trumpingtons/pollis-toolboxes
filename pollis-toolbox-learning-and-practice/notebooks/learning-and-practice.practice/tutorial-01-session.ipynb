# A Practice Session

This notebook goes through one exercise of the Practice panel step by step: the data from a seed, the question and its shuffled choices, working the answer out in Julia, checking it, and new data from another seed.

```julia
using Random, Statistics, Printf, Distributions

fmt(x, digits) = Printf.format(Printf.Format("%.$(digits)f"), x)  # fixed decimals
```

## The Data from a Seed

A random number generator started from a seed makes the data. The same seed gives the same numbers again, so an exercise can be repeated exactly.

```julia
rng = Xoshiro(1)
x = round.(rand(rng, Normal(rand(rng, 10:40), rand(rng, 2:6)), 6); digits = 1)
```

## The Question and the Choices

The choices are computed from the data: the correct one first, then the usual mistakes. Here they are the sample standard deviation, the one that divides by $n$, the variance and the mean absolute deviation.

```julia
question = "A sample of 6 values: $(join(x, ", ")). What is the sample standard deviation?"
choices = [fmt(std(x), 3), fmt(std(x; corrected = false), 3), fmt(var(x), 3), fmt(mean(abs.(x .- mean(x))), 3)]
```

The choices are shown in a shuffled order, so the correct one is not always A. `correct` is where the first choice ended up.

```julia
order = randperm(rng, length(choices))
correct = findfirst(==(1), order)
println(question)
for (letter, i) in zip("ABCD", order)
    println("  ", letter, ") ", choices[i])
end
```

## Working the Answer Out

The sample standard deviation divides the sum of squared deviations by $n - 1$ before the square root. Compute it step by step, then compare it with the choices.

```julia
n = length(x)
deviations = x .- mean(x)
s = sqrt(sum(deviations .^ 2) / (n - 1))
```

## Checking an Answer

Suppose the answer picked is B. Compare its position with `correct`.

```julia
mine = 2  # B
mine == correct ? println("Correct") : println("Not quite: the answer is ", "ABCD"[correct], ") ", choices[1])
```

## New Data

Another seed gives another sample, other choices and another order. Running the same steps for a few seeds shows that the correct answer moves between the letters.

```julia
for seed in 1:5
    rng = Xoshiro(seed)
    x = round.(rand(rng, Normal(rand(rng, 10:40), rand(rng, 2:6)), 6); digits = 1)
    order = randperm(rng, 4)
    println("seed ", seed, ": s = ", fmt(std(x), 3), ", correct choice ", "ABCD"[findfirst(==(1), order)])
end
```
